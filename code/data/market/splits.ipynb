```python
import json
import random

with open("ticker_clusters_2009_2017_train.json", "r") as f:
    clusters = json.load(f)

NUM_RUNS = 50
BATCH_SIZE = 20
BASE_SEED = 42

SELECTED_RUN_IDS = list(range(0, 10)) + list(range(20, 30)) + list(range(40, 50))

random.seed(BASE_SEED)

experiment_plan = []

for run_id in range(NUM_RUNS):

    if run_id < 20:
        subset_type = "high_vol"
        tickers = random.sample(clusters["high_vol"], BATCH_SIZE)

    elif run_id < 40:
        subset_type = "low_vol"
        tickers = random.sample(clusters["low_vol"], BATCH_SIZE)

    else:
        subset_type = "general"
        tickers = random.sample(clusters["general"], BATCH_SIZE)

    experiment_plan.append({
        "run_id": run_id,
        "subset_type": subset_type,
        "tickers": tickers
    })

selected_relabelled_plan = []

for new_run_id, old_run_id in enumerate(SELECTED_RUN_IDS):
    old_row = experiment_plan[old_run_id]

    selected_relabelled_plan.append({
        "run_id": new_run_id,
        "subset_type": old_row["subset_type"],
        "tickers": old_row["tickers"]
    })

with open("experiment_universes_selected_runs.json", "w") as f:
    json.dump(selected_relabelled_plan, f, indent=4)
```
